```python
#Pandas is a Python library used for data manipulation and analysis
import pandas as pd 
```

```python
#load the dataset
df = pd.read_csv('Customer Food Review.csv',index_col='Review_ID')
```

```python
#display the first 5 records
df.head()
```

```python
#number of rows, number of columns
print('Dataset shape:',df.shape)
```

```python
#column names
#data types
#number of non-missing records
#memory usage
print(df.info())
```

```python
#generate descriptive statistics such as count, mean, std, min, etc.
df.describe(include='all')
```

```python
#check missing values
df.isnull().sum()
```

```python
#remove missing values using dropna()
df_drop = df.dropna()
df_drop
```

```python
#fill missing values using fillna()
df_fill = df.copy()
df_fill['Rating'] = df_fill['Rating'].fillna(3)
df_fill
```

```python
df_fill['Comment'] = df_fill['Comment'].fillna('No comment')
df_fill
```

```python
#remove duplicate records
df_unique = df.drop_duplicates()
df_unique
```

```python
#Replace inconsistent text
df['Food'] = df['Food'].replace({
    'burger':'Burger',
    'PIZZA':'Pizza'
})
```

```python
#identify invalid ratings
df[(df['Rating'] < 1) | (df['Rating'] > 5)]
```

```python
#Replace invalid ratings with missing values
df.loc[
    (df['Rating'] < 1) | (df['Rating'] > 5),
    'Rating'
] = None

df
```
